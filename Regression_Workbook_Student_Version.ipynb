In this section, we load the **`house_sales`** dataset from OpenML.

Take a look at the dataset here: [OpenML Dataset 42092 (house_sales)](https://www.openml.org/search?type=data&status=active&id=42092&sort=runs)

The house_sales dataset in OpenML contains real housing transaction records from King County, Washington, including home features and sale prices used for building regression models.

- The answer key also starts by loading this exact dataset.
- After this cell, you will have:
  - `X` for features
  - `y` for the target (house price)
  - `df` for the full DataFrame (features + target)

You can then use these objects throughout the rest of the notebook.

```python
# Load the house_sales dataset from OpenML
openml_id = 42092
target_column = "price"

ds = fetch_openml(data_id=openml_id, as_frame=True)
df = ds.frame.copy()   # full dataframe

# Separate features (X) and target (y)
y = df[target_column]
X = df.drop(columns=[target_column])

print("Shape of X (Total Houses, Input Features):", X.shape)
print("Shape of y (Price Values):", y.shape)

# You can inspect the data in later cells (e.g., df.head()).


```
